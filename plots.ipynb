```python
import torch
import plotly.express as px
```

```python
logprobs_diff = torch.load("results/logprobs_diff.pt").cpu()
mean_logprobs_diff = logprobs_diff.mean(dim=-1).cpu()
```

```python
print("logprobs_diff", logprobs_diff.shape)
print("mean_logprobs_diff", mean_logprobs_diff.shape)
```

```python
# Print the maximum value and its location
max_value = logprobs_diff.max().item()
min_value = logprobs_diff.min().item()
print(f"Min value: {min_value}")
max_index = torch.nonzero(logprobs_diff == max_value, as_tuple=False)
print(f"Max value: {max_value} at index {max_index}")
```

```python
# Plot the max value heatmap of the mean_logprobs_diff
fig = px.imshow(
    mean_logprobs_diff.numpy(),
    labels={"x": "Heads", "y": "Layers", "color": "Mean Logprobs Diff"},
    x=[f"Head {i}" for i in range(mean_logprobs_diff.shape[1])],
    y=[f"Layer {i}" for i in range(mean_logprobs_diff.shape[0])],
    title="Logprobs Difference Heatmap",
    color_continuous_scale="RdBu",
    zmin=-0.5,
    zmax=0.5,
)

fig.show()
```

```python
# Plot the variance heatmap of the logprobs_diff
variance_logprobs_diff = logprobs_diff.var(dim=-1)
fig = px.imshow(
    variance_logprobs_diff.numpy(),
    labels={"x": "Heads", "y": "Layers", "color": "Variance Logprobs Diff"},
    x=[f"Head {i}" for i in range(variance_logprobs_diff.shape[1])],
    y=[f"Layer {i}" for i in range(variance_logprobs_diff.shape[0])],
    title="Logprobs Difference Variance Heatmap",
)
fig.show()
```

```python
# For the first 10 examples, plot the heatmap of logprobs_diff
for i in range(20):
    fig = px.imshow(
        logprobs_diff[:, :, i].numpy(),
        labels={"x": "Heads", "y": "Layers", "color": "Logprobs Diff"},
        x=[f"Head {j}" for j in range(logprobs_diff.shape[1])],
        y=[f"Layer {j}" for j in range(logprobs_diff.shape[0])],
        title=f"Logprobs Difference Heatmap for Example {i}",
        color_continuous_scale="RdBu",
        zmin=-1.0,
        zmax=1.0,
    )

    fig.show()
```
